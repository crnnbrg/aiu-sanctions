# Discipline

```python
import pandas as pd
df = pd.read_csv('data/aiu_ineligible_persons.csv')
```

## Discipline distribution

```python
df.groupby('Country Name')['Discipline 2'].nunique().nlargest(5)
```

```python
df.groupby('Country Name')['Discipline 1'].nunique().nlargest(5)
```

## Athlete Support Personnel

```python
df.query('Role == "Athlete Support Personnel"').groupby('Country Name')['Role'].value_counts().reset_index(name='Count').sort_values('Count', ascending=False)
```

```python
df.query('`Discipline 2` == "Athlete Support Personnel"').groupby('Country Name')['Discipline 2'].value_counts().reset_index(name='Count').sort_values('Count', ascending=False)
```
